```python
import cv2
```

```python
cap =cv2.VideoCapture(0)
i=0
while True:
    flag,frame=cap.read()
    if not flag:
        print("Camera not accessible")
        break
    cv2.imshow("My Camera",frame)
    i+=1
    cv2.imwrite(f"images/img_{i}.jpg",frame)
    key = cv2.waitKey(25) & 0xFF
    #27 -ASCII of escape
    if key == 27:
        break
cap.release()
cv2.destroyAllWindows()
```
