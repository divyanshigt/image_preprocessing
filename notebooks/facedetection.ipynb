```python
import cv2
```

```python
haar_feature = cv2.CascadeClassifier('haarcascade_frontalface_default.xml')
```

```python
img =cv2.imread('all-Indian-cricketer-name.jpg')
```

```python
haar_feature.detectMultiScale(img,scaleFactor=1.1)
```

```python
faces =haar_feature.detectMultiScale(img)
```

```python
for(x,y,w,h) in faces:
    #img -on which we want to draw rectangle
    # (x,y) - top left corner of rectangle
    # (x+w,y+h) - bottom right corner of rectangle
    #(0,255,0) - color of rectangle in BGR format
    #2 - thickness of rectangle border
    cv2.rectangle(img,(x,y),(x+w,y+h),(0,255,0),2)
cv2.imshow('face detection',img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
