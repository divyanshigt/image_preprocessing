```python
import cv2
import yt_dlp
```

```python
URL="https://www.youtube.com/watch?v=wUkABd-g4qg"
#format : 'best[ext=mp4]'
ydl_opts = {
    'format': 'best[ext=mp4]'
}
with yt_dlp.YoutubeDL(ydl_opts) as ydl:
    info=ydl.extract_info(URL,download=False)
    #get the url of video
    video_url=info['url']
    cap=cv2.VideoCapture(video_url)
while True:
    flag,frame=cap.read()
    if not flag:
        print("Video not accessible")
        break
    cv2.imshow("My Video",frame)
    key = cv2.waitKey(25) & 0xFF
    #27 -ASCII of escape
    if key == 27:
        break
cap.release()
cv2.destroyAllWindows()
```
